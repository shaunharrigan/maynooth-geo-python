## **4. Making decisions with conditionals**

An `if` statement runs code **only when a condition is true**. Use `elif` to check another condition and `else` if none of the previous conditions are true.

The diagram below shows how Python checks conditions **in order**, running only the first matching block of code.

<img src="https://raw.githubusercontent.com/shaunharrigan/maynooth-geo-python/main/images/The_Anatomy_of_Conditionals.png" width="900">


### **Comparison operators in Python**

Conditional statements use **comparison operators** to check whether something is `True` or `False`.

| Operator | Meaning | Example |
|:---:|---|---|
| `==` | Equal to | `temperature == 20` |
| `!=` | Not equal to | `temperature != 20` |
| `>` | Greater than | `temperature > 20` |
| `<` | Less than | `temperature < 20` |
| `>=` | Greater than or equal to | `temperature >= 20` |
| `<=` | Less than or equal to | `temperature <= 20` |

**Remember:** `=` assigns a value, while `==` checks whether two values are equal.


### **Demo a simple temperature conditional**

```python
# Demo a few examples
temperature = 31

if temperature >= 30:
    print("Hot day")
elif temperature >= 15:
    print("Mild day")
else:
    print("Cold day")
```

```python
### SOLUTION ###

cities = ["Athens", "Bergen", "Dublin", "Helsinki", "Madrid"]

# Create one figure BEFORE the loop
fig, ax = plt.subplots(figsize=(10, 5))

for city in cities:

    filename = f"era5_{city}_daily.csv"
    df = pd.read_csv(data_dir + filename)

    df["Date"] = pd.to_datetime(df["Date"], format="%d/%m/%Y")
    df = df.set_index("Date")

    annual_temp = df["temperature_mean_DegC"].resample("YS").mean()

    # Add each city's time series to the SAME figure
    ax.plot(annual_temp.index.year, annual_temp.values, label=city)

# Finish the figure AFTER the loop
ax.set_title("Annual mean temperature across Europe")
ax.set_xlabel("Year")
ax.set_ylabel("Annual mean temperature (°C)")
ax.legend()

plt.tight_layout()
fig.savefig(output_dir + "europe_annual_temperature_comparison.png", dpi=150)
plt.close(fig)
```

```python
### ADVANCED SOLUTION: Annual temperature anomalies ###

cities = ["Athens", "Bergen", "Dublin", "Helsinki", "Madrid"]

# Create one figure BEFORE the loop
fig, ax = plt.subplots(figsize=(10, 5))

for city in cities:

    filename = f"era5_{city}_daily.csv"
    df = pd.read_csv(data_dir + filename)

    df["Date"] = pd.to_datetime(df["Date"], format="%d/%m/%Y")
    df = df.set_index("Date")

    # Calculate annual mean temperature
    annual_temp = df["temperature_mean_DegC"].resample("YS").mean()

    # Calculate the 1991–2020 baseline average
    baseline = annual_temp.loc["1991":"2020"].mean()

    # Calculate annual temperature anomalies
    anomalies = annual_temp - baseline

    # Add each city's anomalies to the same figure
    ax.plot(anomalies.index.year, anomalies.values, label=city)

# Finish the figure AFTER the loop
ax.axhline(0, color="black", linestyle="--", linewidth=1)

ax.set_title("Annual temperature anomalies across Europe (1991–2020 baseline)")
ax.set_xlabel("Year")
ax.set_ylabel("Temperature anomaly (°C)")
ax.legend()

plt.tight_layout()
fig.savefig(output_dir + "europe_temperature_anomalies.png", dpi=150)
plt.close(fig)
```
