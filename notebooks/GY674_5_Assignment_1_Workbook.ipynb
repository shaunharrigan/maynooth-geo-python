### Dataset background and variable reference

The **Pobal HP Deprivation Index** measures relative socioeconomic disadvantage across Ireland using information from the national census. It combines indicators relating to education, employment, social class and demographic characteristics.

The dataset contains information for individual **Electoral Divisions (EDs)**, allowing us to investigate geographical differences both between and within counties.

The main variable used in this assignment is `Index22_ED_std_rel_wt`, the **2022 standardised relative deprivation score**. Lower scores indicate greater deprivation, while higher scores indicate greater affluence.

**Further information:**
- [Pobal HP Deprivation Index](https://www.pobal.ie/pobal-hp-deprivation-index/)
