# **Step 3. PREPARE**

Our first map revealed three things to consider before analysing the data:

- **Variable:** Select the 2-metre air temperature variable (`t2m`) from our dataset.
- **Temperature:** Convert from Kelvin to degrees Celsius (°C).
- **Longitude:** Change from 0–360° to −180–180° for a more familiar world map.

👀 **Demo. Prepare our temperature data**

We'll select the variable, make both adjustments, and save the result as an xarray **DataArray** called `temp`.



👀 **Demo. Prep step 1: Convert Kelvin to degrees Celsius**

ERA5 temperature is stored in **Kelvin**. Subtract 273.15 to get °C. Keep the result as an xarray **DataArray**.

```python
# Select 2-metre air temperature and convert Kelvin to Celsius
temp = ds["t2m"] - 273.15

# Update the units
temp.attrs["units"] = "°C"
```


👀 **Demo. Prep step 2: Adjust longitude coordinates**

ERA5 uses longitudes from **0–360°**. Convert these to **−180° to 180°** for a more familiar world map.

Use `.assign_coords()` to update the coordinates and `.sortby()` to put them in the correct order.

```python
# Convert longitude from 0–360° to -180–180°
lon = (temp.longitude + 180) % 360 - 180

# Update longitude coordinates and sort west to east
temp = temp.assign_coords(longitude=lon).sortby("longitude")
```
