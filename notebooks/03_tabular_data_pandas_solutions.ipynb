## Before we start | What is tabular data?

**Tabular data** is information arranged in a table:
- **Rows** represent observations (e.g. one weather station or one geographical area).
- **Columns** represent variables (e.g. station name, rainfall, population).
- **Values** are the entries in the table.

A **CSV** (*comma-separated values*) is a common file format for storing tables.

**Pandas** is a Python package for working with tables. In Pandas, a table is called a **DataFrame**.



### 👀 Demo | Create a subset of our DataFrame

Our original DataFrame, `df`, contains all the variables. We only need six of them for our analysis.

We'll create a **subset** called `df_sub`, containing just the columns we want:

- `df[[...]]` selects the columns.
- `df_sub =` gives our subset a new name.
- `.copy()` creates an independent copy, so changes to `df_sub` won't affect `df`.

**The six columns to select:**

- `"COUNTY"`
- `"ED_ENGLISH"`
- `"UNEMPM22"`
- `"UNEMPF22"`
- `"EDHIGH22"`
- `"Index22_ED_std_rel_wt"`

**Important:** We'll keep the original `df` unchanged and use `df_sub` for our analysis.

**Remember:** Subsetting means selecting part of a larger dataset. We can subset by **columns** (variables), **rows** (observations), or both.


```python
# subset six columns (variables) we want to do more analysis on...

df_sub = df[[
    "COUNTY",
    "ED_ENGLISH",
    "UNEMPM22",
    "UNEMPF22",
    "EDHIGH22",
    "Index22_ED_std_rel_wt"
]].copy()
```

```python
df_sub
```


### 👀 Demo | Rename columns

Our subset, `df_sub`, now contains the six variables we need. However, some column names are difficult to understand.

We can use **`.rename()`** to give columns clearer, more meaningful names.

- `.rename(columns={...})` tells Pandas which column names to change.
- Each pair contains the **original name** followed by the **new name**.
- `df_sub =` saves the renamed DataFrame back to the same variable.

**Column names to rename (copy during the demo):**

- `"ED_ENGLISH": "ED"`
- `"UNEMPM22": "Male_unemployment_%"`
- `"UNEMPF22": "Female_unemployment_%"`
- `"EDHIGH22": "Higher_education_%"`
- `"Index22_ED_std_rel_wt": "Deprivation_score"`

We'll leave `COUNTY` unchanged.

**Remember:** Renaming a column changes its label, not the data it contains.


```python
# rename columns
df_sub = df_sub.rename(columns={
  "ED_ENGLISH": "ED",
  "UNEMPM22": "Male_unemployment_%",
  "UNEMPF22": "Female_unemployment_%",
  "EDHIGH22": "Higher_education_%",
  "Index22_ED_std_rel_wt": "Deprivation_score"
  })
```



### 👀 Demo | Filter rows

So far, we've selected **columns** to create our subset. Now let's select **rows**.

**Filtering** means keeping only the rows that meet a particular condition.

For example, let's filter our data to show only Donegal, keeping all six variables.

Here's how it works:

- `df_sub["COUNTY"]` selects the county column.
- `== "DONEGAL"` checks which rows match Donegal (`True` or `False`).
- `df_sub[...]` keeps only the matching rows.
- `df_sub_donegal =` saves the filtered data as a new DataFrame.

**Remember:**
- `==` means **is equal to** (a comparison).
- `=` means **assign a value**.
- Selecting **columns** reduces the variables; filtering **rows** reduces the observations.

Our original `df_sub` remains unchanged.

```python
df_sub["COUNTY"] == "DONEGAL"
```

```python
df_sub_donegal = df_sub[df_sub["COUNTY"] == "DONEGAL"].copy()
df_sub_donegal
```

## Understanding `groupby()`

**`.groupby()`** lets us group data by a category (e.g. county) and calculate statistics for each group.

Imagine we have four observations from two counties:

| County | Higher education (%) |
|---|---:|
| Donegal | 20 |
| Donegal | 30 |
| Kildare | 40 |
| Kildare | 50 |

If we **group by county** and calculate the **mean**:

- **Donegal:** (20 + 30) / 2 = **25%**
- **Kildare:** (40 + 50) / 2 = **45%**

We've reduced **four observations to two county summaries**.

### 👀 Demo | Group a small table first

We'll use `.groupby("County")` to group the observations, then `.mean()` to calculate the average higher education percentage for each county.


```python
# Create a small example dataset
small = pd.DataFrame({
    "COUNTY": ["Donegal", "Donegal", "Kildare", "Kildare"],
    "Higher_education_%": [20, 30, 40, 50]
})

# Group by county and calculate the mean
small.groupby("COUNTY")["Higher_education_%"].mean()
```


### 👀 Demo | Calculate county averages

- `.groupby("COUNTY")` — groups the rows by county.
- `["Higher_education_%"]` — selects the variable.
- `.mean()` — calculates the average for each county.

### Your task

Starting with `df_sub`:

1. Use `groupby()`, `.mean()` and `.sort_values()` to calculate and rank mean deprivation scores by county.
2. Create a **horizontal bar chart** comparing all counties.
3. Give your chart a clear title and labelled axes.
4. Write **one or two sentences** answering the research questions.

**Hints:**
- Use the `Deprivation_score` column.
- Adapt the higher education analysis from Steps 4 and 5.
- For a horizontal bar chart, use `.plot.barh()` instead of `.plot.bar()`.

**Remember:** More negative scores indicate greater relative disadvantage; more positive scores indicate greater relative affluence. These are unweighted averages of ED scores, not official county deprivation scores.


```python
# Part 1 | Investigate deprivation by county

# Calculate mean deprivation score by county
df_sub_county_deprivation = df_sub.groupby("COUNTY")["Deprivation_score"].mean()

# Rank counties from most disadvantaged to most affluent
df_sub_county_deprivation.sort_values()
```

```python
# Lowest mean score = most disadvantaged
print("Most disadvantaged county:")
print(df_sub_county_deprivation.idxmin())

# Highest mean score = most affluent
print("Most affluent county:")
print(df_sub_county_deprivation.idxmax())
```

```python
# Visualise mean deprivation scores across counties
df_sub_county_deprivation.sort_values().plot.barh(
    figsize=(9, 7),
    title="Mean deprivation score by county (2022)",
    xlabel="Mean deprivation score",
    ylabel="County"
)
```

```python
# Red colours for negative, blue for postitive deprivation scores
# Sort county deprivation scores
scores = df_sub_county_deprivation.sort_values()

# Red = negative, blue = positive
colours = ["tab:red" if score < 0 else "tab:blue" for score in scores]

scores.plot.barh(
    figsize=(9, 7),
    color=colours,
    title="Mean deprivation score by county (2022)",
    xlabel="Mean deprivation score",
    ylabel="County"
)
```
