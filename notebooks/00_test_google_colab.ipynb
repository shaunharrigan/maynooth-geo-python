## 2. Try a simple calculation

Python can store values and use them in calculations. This example converts a temperature from kelvin to degrees Celsius.

```python
temperature_k = 291.55  # Temperature in kelvin
temperature_c = temperature_k - 273.15  # Convert to degrees Celsius

print("Temperature:", round(temperature_c, 1), "°C")
```

The result should be `18.4 °C`.

## 3. Check the main course packages

A Python **package** adds tools for particular types of work. Run this cell to confirm that the principal packages used in the course are available.

```python
import numpy as np               # Numerical calculations and arrays
import pandas as pd              # Tabular data analysis
import xarray as xr              # Multidimensional and gridded data
import matplotlib.pyplot as plt  # Data visualisation

print("NumPy version:", np.__version__)
print("Pandas version:", pd.__version__)
print("Xarray version:", xr.__version__)
print("Matplotlib is available")
```

Version numbers may differ over time. If the cell runs without an error, the packages are available.

## 4. Create a simple figure

Run the final code cell to check that Colab can create and display a figure.

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
temperature = [5.2, 5.6, 7.4, 9.1, 12.0]

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(months, temperature, marker="o", color="firebrick")
ax.set(
    title="Example monthly temperature",
    xlabel="Month",
    ylabel="Temperature (°C)"
)
ax.grid(alpha=0.25)
plt.show()
```
